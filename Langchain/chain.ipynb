```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
llm = ChatGoogleGenerativeAI(
    model="gemini-2.5-flash", 
    temperature=0, 
    api_key="AIzaSyDO4GsJio6Flrs5gxOezoSfuqvGI_EKpFs"
)
```

```python
parser = StrOutputParser()
```

Translate chain

```python
translate_prompt = ChatPromptTemplate.from_messages([
    ('system','You are a helpful assistant,Dectect the User Language and Translate it to English '),
    ('human','{prompt}')
])
translate_chain = translate_prompt | llm | parser
```

Summarize

```python
summarize_prompt = ChatPromptTemplate.from_messages([
    ('system','You are a helpful assistant,Summarize the Text '),
    ('human','{prompt}')
])
summarize_chain = summarize_prompt | llm | parser
```

```python
full_chain = translate_chain | summarize_chain
full_chain1 = ({'prompt' :translate_chain } | summarize_chain)
result = full_chain.invoke({'prompt':'''Yenakku udambu seri illa 
                   athanala naan office pola'''})
```

```python
full_chain1 = ({'prompt' :translate_chain } | summarize_chain)
result = full_chain1.stream({'prompt':'''Yenakku udambu seri illa 
                   athanala naan office pola'''})
for chunk in result:
    print(chunk)
```
