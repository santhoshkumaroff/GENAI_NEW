'''
POS - Part of speech

The Part Of Speech tag. Valid options are "n" for nouns, "v" for verbs, "a" for adjectives, "r" for adverbs
'''

```python
import nltk
''' 
syntax

nltk.pos_tag(tokens)
'''
words = ['MEDIUM','playing','football','studied','training','plays','locals']

tags = nltk.pos_tag(words)
```

```python
def pos_tagger(tag):
    if tag.startswith('N'):
        return 'n'
    elif tag.startswith('J'):
        return 'a'
    elif tag.startswith('V'):
        return 'v'
    elif tag.startswith('R'):
        return 'r'
    else:
        return 'n'
words = ['MEDIUM','playing','football','studied','training','plays','locals']
tags = nltk.pos_tag(words)
for word,tag in tags:
    print(lemma.lemmatize(word,pos_tagger(tag)))
```
